# Langkah 1: Menangani Nilai Hilang (imputasi media/modus)

```python
import pandas as pd
import numpy as np
data_Karyawan= {
    'usia': [25, 30, np.nan, 45, 35, 50, np.nan, 28],
    'pendapatan': [5000, 7000, 6000, np.nan, 8000, 12000, 9000, 5500],
    'pendidikan': ['SMA', 'S1', 'S2', 'SMA', 'S1', 'S2', 'S1', 'SMA'],
    'kota': ['Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung'],
    'status_pernikahan': ['Lajang', 'Menikah', 'Lajang', 'Menikah', 'Menikah', 'Lajang', 'Menikah', 'Lajang'], # Latihan 3
    'membeli': ['Tidak', 'Ya', 'Tidak', 'Ya', 'Ya', 'Ya', 'Tidak', 'Tidak']
    }
df = pd.DataFrame(data_Karyawan)
```

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

## Langkah 2 : Memisahkan fitur (x) dan label (y)

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

# Langkah 3: Encoding kategorikal(ordinal & one-hot)

```python
X['pendidikan'] = X['pendidikan'].map({'SMA': 0, 'S1': 1, 'S2': 2})
X = pd.get_dummies(X, columns=['kota'], dtype=int)
y = y.map({'Tidak': 0, 'Ya': 1})
```

# Langkah 4: Membagi data (sebelum scaling)

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

# Langkah 5: Penskalaan

```python
from sklearn.preprocessing import StandardScaler
num = ['usia', 'pendapatan', 'pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])
```

# Langkah 6: Verifikasi

```python
# 1. Cek apakah masih ada nilai hilang (missing values)
print("=== 1. Cek Missing Values ===")
print(X_train.isnull().sum())
print("\n" + "="*40 + "\n")

# 2. Cek tipe data (memastikan semua kolom sudah bertipe numerik)
print("=== 2. Cek Tipe Data Kolom ===")
print(X_train.dtypes)

print("\n" + "="*40 + "\n")

# 3. Cek statistik deskriptif (memastikan skala fitur numerik sudah seragam)
print("=== 3. Cek Skala Fitur (Statistik Deskriptif) ===")
print(X_train.describe())
```

# 4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.

fit_transform() pada data latih menghitung parameter penskalaan (seperti nilai min, max pada MinMaxScaler atau mean, std pada StandardScaler) sekaligus mengaplikasikannya.   transform() pada data uji hanya menggunakan parameter statistik yang sudah dihitung dari data latih. Hal ini dilakukan untuk mensimulasikan kondisi dunia nyata di mana data uji dianggap sebagai data masa depan yang belum pernah dilihat (unseen data).

# E. Refleksi

1. Mengapa urutan 'split dulu, baru scaling' penting?
Untuk mencegah terjadinya Data Leakage (kebocoran data). Jika penskalaan dilakukan sebelum split, nilai statistik seluruh dataset (termasuk data uji) akan memengaruhi proses penskalaan data latih. Hal ini membuat model "mencuri start" dengan mengetahui informasi tentang distribusi data uji, sehingga evaluasi model menjadi tidak objektif/terlalu optimis.
2. Kapan label encoding dan kapan one-hot encoding?
Label / Ordinal Encoding: Digunakan ketika fitur kategorikal memiliki tingkatan/urutan logis (misal: SMA < S1 < S2 atau Rendah < Sedang < Tinggi).   One-Hot Encoding: Digunakan ketika fitur kategorikal bersifat nominal (tidak memiliki urutan) (misal: nama kota Jakarta, Bandung, Surabaya).
3. Apa perbedaan normalisasi dan standardisasi?
Normalisasi (MinMaxScaler): Mengubah skala data ke rentang tertentu (biasanya 0 hingga 1). Sangat berguna jika algoritma tidak mengasumsikan distribusi data tertentu (seperti Neural Networks atau KNN).   Standardisasi (StandardScaler): Mengubah data agar memiliki nilai rata-rata ($\mu = 0$) dan standar deviasi ($\sigma = 1$). Sangat cocok untuk data yang berdistribusi normal/Gaussian dan algoritma yang sensitif terhadap skala seperti SVM, Logistic Regression, atau PCA
